```python
import tensorflow as tf
import numpy as np

# Training data
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

# Create model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

# Compile model
model.compile(
    optimizer='sgd',
    loss='mean_squared_error'
)

# Train model
model.fit(x, y, epochs=500, verbose=0)

# Predict
result = model.predict(np.array([10]))
print("Prediction for 10:", result)
```

```python
import tensorflow as tf
import numpy as np

# 1. Training data
texts = tf.constant([
    "I love this product",
    "This is amazing",
    "I am very happy",
    "The movie was excellent",
    "I hate this",
    "This is terrible",
    "I am very sad",
    "The movie was bad"
])

labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])

# 2. Convert text into numbers
vectorizer = tf.keras.layers.TextVectorization(
    max_tokens=1000,
    output_sequence_length=10
)

vectorizer.adapt(texts)

# 3. Build AI model
model = tf.keras.Sequential([
    vectorizer,
    tf.keras.layers.Embedding(input_dim=1000, output_dim=16),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

# 4. Compile model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 5. Train model
model.fit(texts, labels, epochs=30)

# 6. Test new sentence
test_sentences = tf.constant([
    "I really love this",
    "This is very bad",
    "I am happy",
    "I hate the movie"
])

predictions = model.predict(test_sentences)

for sentence, prediction in zip(test_sentences.numpy(), predictions):
    score = prediction[0]

    if score >= 0.5:
        sentiment = "Positive"
    else:
        sentiment = "Negative"

    print(sentence.decode(), "=>", sentiment, "Score:", round(score, 2))
```
