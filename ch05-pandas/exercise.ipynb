```python
import pandas as pd
data = {
    "Name": ["Ali", "Siti", "John", "Ali"],
    "Subject": ["Math", "Math", "Math", "Science"],
    "Marks": [85, 90, 88, 80]
}

df = pd.DataFrame(data)

grouped = df.groupby("Name")["Marks"].mean()
print(grouped)
```
