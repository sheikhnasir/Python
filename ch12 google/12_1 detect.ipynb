```python
from textblob import TextBlob

text = input("Enter your sentence: ")

analysis = TextBlob(text)

polarity = analysis.sentiment.polarity

if polarity > 0:
    sentiment = "Positive"
elif polarity < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print("Text:", text)
print("Polarity Score:", polarity)
print("Sentiment:", sentiment)
```
